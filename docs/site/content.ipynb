# The content of a slide

Every cell below holds the Markdown of one deck and nothing else; the deck plays under it. Open the notebook live ("Open in JupyterLite" above), change a cell, run it again, and watch the deck follow. The first cell loads the helper; from then on a cell that starts with `%%deck` is a deck.

This notebook is about what a slide can hold. Each cell adds one kind of content; the pages that follow it in the sidebar give every detail.

````python
import zoomdeck_nb
````

## Prose

A slide holds Markdown prose: paragraphs, emphasis, inline code and inline math, coloured spans in braces. Text flows from the top left of the canvas and wraps to the slide's width; `{#id x= y= width=}` after a paragraph places it instead.

````python
%%deck
::: slide {#prose width=20}
## Prose

Paragraphs wrap to the canvas. Use **bold**, *italic*, ***both***, `inline code` and inline math $e^{i\pi} + 1 = 0$.

[Coloured spans]{color=ORANGE} take palette names, [CSS colours]{style="color:#4ADE80"} or hex values, and [nest **emphasis** and $\sqrt{2}$]{color=TEAL}.

A paragraph placed by hand sits where its attributes say. {#aside x=2 y=-2 width=7 font_size=20 color=GRAY}
:::
````

## Headings and lists

Headings set the size only (`#` to `######`); nesting between slides comes from includes, never from headings. Lists nest and wrap with a hanging indent.

````python
%%deck
::: slide {#lists width=20}
## A heading

### A smaller one {color=TEAL}

- Bullet items are prose with a marker
  - Nested items indent
    1. Numbered lists nest too
    2. And count on
- Links like [this one](https://puppis.github.io/zoomdeck/) are clickable

First line.\\
Second line after an explicit break.
:::
````

## Formulas

Inline math shares the baseline of its line; a `$$` block is a display formula, centred. A formula may be an element of its own, with a position, a size and a step.

````python
%%deck
::: slide {#math width=20}
## Formulas

Inline math shares the baseline: $\int_0^1 x^2\,dx = \tfrac13$, and so does $\binom{n}{k}$.

$$
\sum_{k=0}^{n} \binom{n}{k} = 2^n
$$

::: {.element #pythagoras x=0.6 y=-2.6 font_size=34 step=1}
$$ \textcolor{Orange}{a}^2 + {\color{Cerulean} b^2} = c^2 $$
:::
:::
````

## Listings

A fenced block is a code listing, highlighted by language, placed like any element with attributes after the language.

````python
%%deck
::: slide {#code width=20}
## A listing

```python {#fib x=-9 y=3 width=9}
def fib(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

The listing on the left is placed at `x=-9 y=3`, nine units wide; its language picks the colours. {#note x=1 y=3 width=8 font_size=20}
:::
````

## Tables

A Markdown table becomes a table element; the alignment row sets the columns, `rules=` and `stripe_opacity=` its look, and cells may hold emphasis and math.

````python
%%deck
::: slide {#tables width=20}
## A table

::: {.element #sales x=0.6 y=1.6 rules=rows stripe_opacity=0.08}
| Region | 2023 | 2024 |
| :--- | ---: | ---: |
| North | 3.2 | **5.0** |
| South | 2.1 | $\pi$ |
| East | 1.4 | 1.9 |
:::
:::
````

## Images

An image is Markdown's `![alt](url)` with a size; SVG, PNG, JPEG and animated PNG or GIF all work, and a video is an image with a `.mp4` or `.webm` source. Here the picture comes from the documentation site.

````python
%%deck
::: slide {#images width=20}
## An image

![An orbit](https://puppis.github.io/zoomdeck/live/orbit.png){#orbit x=-6 y=-0.6 width=5}

Images are placed by their centre; `width` or `height` sizes them, the other follows the picture's proportions. {#cap x=0 y=1.2 width=8 font_size=20}
:::
````
