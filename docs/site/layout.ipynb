# Layout and style

Every cell below holds the Markdown of one deck and nothing else; the deck plays under it. Open the notebook live ("Open in JupyterLite" above), change a cell, run it again, and watch the deck follow. The first cell loads the helper; from then on a cell that starts with `%%deck` is a deck.

This notebook is about placing things on a slide and giving them a common look. Coordinates are the slide's own: the origin at the centre of the canvas, `x` to the right, `y` up, one unit the same at every zoom.

```python
import zoomdeck_nb
```

## Flow and positions

Without attributes, elements flow from the top left. `x`, `y` place an element instead: text by its left edge and first baseline, shapes and images by their centre.

```python
%%deck
::: slide {#positions width=20}
## Flow and positions

This paragraph flows at the top.

So does this one, under it.

[Placed at x=2, y=-1]{#p x=2 y=-1 font_size=24}

[[dot: #origin x=0 y=0 color=RED]]

[the origin]{at=#origin.right anchor=left-baseline dx=0.2 font_size=16 color=GRAY}

[[rectangle: #r x=-6 y=-3 width=3 height=1.4 color=BLUE]]
:::
```

## Anchors

`anchor=` chooses which point of an element the position refers to: `top-left`, `bottom-right`, `center`, `right-baseline` and so on, so that things align on edges rather than on their defaults.

```python
%%deck
::: slide {#anchors width=20}
## Anchors

[[rectangle: #frame x=0 y=-0.5 width=16 height=6 color=GRAY stroke_width=1]]

[Top left of the frame]{#tl x=-8 y=2.5 anchor=top-left font_size=22}

[Ends at the right edge]{#rb x=8 y=-3 anchor=bottom-right font_size=22}

[[circle: #c x=8 y=2.5 radius=0.6 color=TEAL anchor=top-right]]

[Centred]{#cc x=0 y=-0.5 anchor=center font_size=30 color=TEAL}
:::
```

## Relative constraints

An element can be placed relative to another: `below:#id`, `right-of:#id`, `left-aligned:#id`, `baseline-aligned:#id`, with `buff` as the distance; `at=` puts one of its points on a point of another element.

```python
%%deck
::: slide {#relative width=20}
## Relative constraints

### A heading {#h x=-8 y=3}

[Below it, left-aligned]{#p below:#h left-aligned:#h buff=0.3}

[A side note]{#note right-of:#p buff=0.5 font_size=18 color=GRAY}

[Big]{#big x=-8 y=-1.5 font_size=48}

[shares the baseline]{x=-5.5 baseline-aligned:#big}

[[rectangle: #ref x=4 y=-2.5 width=3 height=1.6 color=BLUE]]

[My top left at its bottom right]{#tl at=top-left:#ref.bottom-right dx=0.2 font_size=16}

[[dot: #d at=#ref.top color=RED]]
:::
```

## Groups

A `::: {.group #id}` container places several elements together, with coordinates of their own inside it; the group is then one element for positions, steps and enclosures.

```python
%%deck
::: slide {#groups width=20}
## Groups

::: {.group #automaton x=-3 y=0.5 step=1}
[[circle: #q0 x=0 y=0 radius=0.6 color=BLUE]]
[$q_0$]{at=#q0.center anchor=center}
[[circle: #q1 x=3 y=0 radius=0.6 color=BLUE]]
[$q_1$]{at=#q1.center anchor=center}
[[arrow: #q0 -> #q1, #t, label="$a$"]]
:::

[[box: #automaton, #gbox, buff=0.3, color=GREEN, step=2]]

::: {.group #legend below:#automaton left-aligned:#automaton buff=0.9}
[[dot: #k1 x=0 y=0 color=BLUE]]
[a state]{at=#k1.right anchor=left-baseline dx=0.2}
:::
:::
```

## Styles

A `::: styles` block names styles once: by role (`heading`, `body`, `shape`, `icon`), or by a `.class` that elements pick up. The same block can live in `deck.json` for a whole deck.

```python
%%deck
::: styles
heading: color=BLUE_E
callout: color=ORANGE font_size=30
footnote: font_size=16 color=GRAY
:::

::: slide {#styles width=20}
## A styled heading

Body text as usual. {#body x=-8 y=2}

[Look here]{#c .callout below:#body left-aligned:#body buff=0.6}

[A side remark in the footnote style]{.footnote below:#c left-aligned:#c buff=0.6}
:::
```
