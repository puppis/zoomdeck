# Diagrams

Every cell below holds the Markdown of one deck and nothing else; the deck plays under it. Open the notebook live ("Open in JupyterLite" above), change a cell, run it again, and watch the deck follow. The first cell loads the helper; from then on a cell that starts with `%%deck` is a deck.

Diagrams are directives in double brackets: a shape, an arrow between two elements, an icon, an enclosure around several of them. Each cell adds one kind.

```python
import zoomdeck_nb
```

## Shapes

A shape directive names the kind, gives it an `#id`, a position and a size, then its colours. Shapes are placed by their centre.

```python
%%deck
::: slide {#shapes width=20}
## Shapes

[[rectangle: #box x=-6 y=0 width=3.4 height=1.8 corner_radius=0.3 color=BLUE fill_color=BLUE fill_opacity=0.15]]

[[circle: #disk x=-1 y=0 radius=0.9 color=GREEN stroke_width=4]]

[[ellipse: #e x=3.5 y=0 width=3.4 height=1.6 color=TEAL fill_color=TEAL fill_opacity=0.2]]

[[polygon: #tri points="(-7,-4) (-5,-2) (-4,-4)" color=ORANGE fill_color=ORANGE fill_opacity=0.3]]

[[arc: #a x=0 y=-3 radius=1.4 start_angle=0 angle=2.4 color=RED stroke_width=4]]

[[polyline: #zigzag points="(3,-4.2) (5,-2.3) (7,-3.8) (9,-2)" color=ORANGE stroke_width=3]]
:::
```

## Arrows and lines

An arrow or a line joins two endpoints: elements by `#id`, or points in parentheses. A label sits along it, and `step=` makes it appear on a click.

```python
%%deck
::: slide {#arrows width=20}
## Arrows and lines

[[rectangle: #box x=-6 y=0 width=3.4 height=1.8 corner_radius=0.3 color=BLUE fill_color=BLUE fill_opacity=0.15]]

[[circle: #disk x=2 y=0 radius=0.9 color=GREEN stroke_width=4]]

[[arrow: #box -> #disk, #edge, label="**labelled** $\to$", color=GOLD_E]]

[[line: #disk -> (6,-3.5), #drop, color=GRAY, stroke_width=2]]

[[arrow: (-8.8,-3.5) -> #box, #pointer, color=RED, step=1]]
:::
```

## Icons

Icons come from a built-in set (`user`, `server`, `database`, `cloud`, `document` and more); a label sits beside them, and they take the same colours and arrows as shapes.

```python
%%deck
::: slide {#icons width=20}
## Icons

[[icon: user #alice x=-7 y=-0.5 size=1.4 color=BLUE_E label="Alice"]]

[[icon: server #api right-of:#alice buff=3 size=1.4 color=GREEN_E fill_opacity=0.25 label="api" label_side=above]]

[[icon: database #db right-of:#api buff=3 size=1.4 color=GOLD_E fill_opacity=0.3 label="orders"]]

[[icon: cloud #cdn right-of:#db buff=2 size=1.4 color=TEAL_E label="cdn"]]

[[arrow: #alice -> #api, #req, label="HTTPS"]]

[[arrow: #api -> #db, #sql, label="SQL", step=1]]
:::
```

## Boxes, braces and blobs

An enclosure lists the elements it surrounds before the first comma, then its own `#id`: a `box` is their joint rectangle grown by `buff`, a `blob` a smooth curve around them, a `brace` sits on one side of one element.

```python
%%deck
::: slide {#enclosures width=20}
## Boxes, braces and blobs

[[icon: server #web x=-6.5 y=0.3 size=1.1 color=GREEN_E label="web" label_side=right]]

[[icon: server #worker x=-6.5 y=-1.7 size=1.1 color=GREEN_E label="worker" label_side=right]]

[[icon: database #store x=-1.5 y=-0.7 size=1.1 color=GOLD_E label="store" label_side=right]]

[[box: #web #worker, #tier, buff=0.3, corner_radius=0.2, color=GRAY, step=1]]

[[blob: #web #worker #store, #backend, buff=0.45, smoothness=0.7, color=GREEN_E, fill_opacity=0.08, step=2]]

[[brace: #store, #bs, side=bottom, label="the data", color=GRAY, step=3]]
:::
```

## Hand-drawn

`sketch=true` draws a shape, an icon, an arrow or an enclosure as if by hand: a wobbly outline and a hatched or solid fill. A `::: styles` block turns it on for whole kinds at once.

```python
%%deck
::: styles
shape: sketch=true roughness=1.2
connection: sketch=true
attachment: sketch=true
:::

::: slide {#sketchy width=20}
## Hand-drawn

[[rectangle: #hr x=-6.5 y=-0.5 width=3.4 height=1.8 color=BLUE_E fill_opacity=0.5]]

[[circle: #hc right-of:#hr buff=1.5 radius=0.9 color=GREEN_E fill_opacity=0.6 fill_style=cross-hatch]]

[[polygon: #hp points="(1.5,-1.4) (3,0.5) (4.5,-1.4)" color=ORANGE fill_opacity=1 fill_style=solid roughness=2]]

[[arrow: #hr -> #hc, #ha]]

[[brace: #hp, #hb, side=bottom, label="by hand"]]

[[rectangle: #exact x=7 y=-0.5 width=2.4 height=1.4 color=GRAY sketch=false]]
:::
```
