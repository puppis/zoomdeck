# Motion and interaction

Every cell below holds the Markdown of one deck and nothing else; the deck plays under it. Open the notebook live ("Open in JupyterLite" above), change a cell, run it again, and watch the deck follow. The first cell loads the helper; from then on a cell that starts with `%%deck` is a deck.

A slide plays in steps: each click reveals, hides, moves or transforms something. This notebook adds one kind of motion per cell.

```python
import zoomdeck_nb
```

## Steps

`step=N` reveals an element on the N-th click, `hide=N` hides it; the element is otherwise there from the start. `[[show:]]` and `[[hide:]]` change visibility again later.

```python
%%deck
::: slide {#steps width=20}
## Steps

[Visible on entry, hidden at step 3]{#hint x=-8 y=2 hide=3}

[Appears at step 1]{#one x=-8 y=1 step=1 color=GREEN}

[Appears at step 2]{#two x=-8 y=0 step=2 color=GOLD_E}

[Step 3 shows this and hides the hint]{#three x=-8 y=-1 step=3}

[[show: #hint step=4]]

[[hide: #one step=4]]
:::
```

## Timing and automatic advancement

`run_time` and `lag` time a reveal; `[[next: step=N]]` advances to the following stop by itself once step N has played, and `[[skip_next: step=N]]` leaves exactly one following event out.

```python
%%deck
::: slide {#timing width=20}
## Timing

[Fades in over 1.5 s]{#slow x=-8 y=1.5 step=1 run_time=1.5}

[Starts 0.5 s later]{#later below:#slow left-aligned:#slow step=1 lag=0.5}

[Step 2 reveals this, then advances by itself]{#auto x=-8 y=-1 step=2 color=GOLD_E}

[[next: step=2]]

[Step 3 follows at once]{#three x=-8 y=-2 step=3}

[[skip_next: step=3]]

[Never shown: step 4 is skipped]{#four x=-8 y=-3 step=4}

[Step 5 follows step 3 directly]{#five x=-8 y=-4 step=5 color=RED}
:::
```

## Animations

`[[animate: #id, step=N, …]]` moves, turns, scales or recolours an element; `[[effect: name #id]]` plays a named effect, `[[transform: #a -> #b]]` morphs one element into another, and `enter=`/`exit=` choose how an element appears or leaves; `enter=write` writes it stroke by stroke, here in the bundled handwriting face (`font=Zeyada`).

```python
%%deck
::: slide {#animations width=20}
## Animations

[[circle: #ball x=-7 y=1.5 radius=0.6 color=BLUE fill_color=BLUE fill_opacity=0.6]]

[[animate: #ball, step=1, x=7, run_time=1.2, rate=ease_in_out]]

[[rectangle: #card x=-7 y=-1.5 width=2.6 height=1.6 color=GREEN fill_color=GREEN fill_opacity=0.2]]

[[animate: #card, step=2, angle=0.4, scale=1.3, color=ORANGE, opacity=0.6]]

[Before]{#before x=0 y=-1.6 font_size=30}

[[effect: indicate #before, step=3, color=GOLD_E]]

[After]{#after x=0 y=-1.6 font_size=30 color=TEAL step=4}

[[transform: #before -> #after, step=4, run_time=1]]

[Written by hand]{#hand x=0 y=-4 font=Zeyada font_size=44 enter=write step=5}
:::
```

## Focus-only decorations

`focus_only=true` marks an element that is visible only while the camera rests on its slide: it fades in with the focus and out when the camera leaves, so parents show their children's content without the decoration.

```python
%%deck
::: slide {#focus width=20}
## Focus-only decorations

[[rectangle: #paper x=0 y=0 width=20 height=11.25 fill_color=TEAL fill_opacity=0.12 stroke_width=0 focus_only=true]]

[This caption fades in with the focus and out when the camera leaves]{#caption x=-9 y=0.6 width=18 focus_only=true color=TEAL}

[This ordinary text stays visible from the parent too]{#plain x=-9 y=-2 width=18}

[[rectangle: #inner-frame x=5 y=-3.5 width=6 height=3.38 color=GRAY stroke_width=1]]

[[include: #inner x=5 y=-3.5 width=6 preview=true]]
:::

::: slide {#inner}
## Looking back

From here the parent's tinted paper and its caption are gone. {font_size=22}
:::
```

## Links

`href=` on any element makes it a link: to a web page, or to a stop of the deck with `#slide=id` or `#step=N`. The front page of the tour is built this way.

```python
%%deck
::: slide {#links width=20}
## Links

[The documentation ↗]{#doc x=-8 y=1.5 font_size=28 color=BLUE_E href=https://puppis.github.io/zoomdeck/}

[Jump to the second slide]{#jump x=-8 y=0 font_size=28 color=BLUE_E href=#slide=second}

[[rectangle: #hit x=4 y=0 width=6 height=3 color=GRAY href=#slide=second]]

[a shape can be a link too]{#hitcap at=#hit.center anchor=center font_size=18 color=GRAY}

[[rectangle: #second-frame x=4 y=-4 width=5 height=2.81 color=GRAY stroke_width=1]]

[[include: #second x=4 y=-4 width=5 preview=true]]
:::

::: slide {#second}
## The second slide

[Back to the first]{#back x=-7 y=0 font_size=28 color=BLUE_E href=#slide=links}
:::
```
