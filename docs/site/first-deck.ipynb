# Your first deck, stage by stage

This notebook builds a deck one cell at a time. Each cell holds the Markdown of a deck, nothing else, and shows the deck below it, played by the real player: click the slides and press the arrow keys, or use the buttons under them. Open the notebook live ("Open in JupyterLite" above) to change any cell and run it again; the deck below it follows.

The first cell loads the helper. From then on a cell that starts with `%%deck` is a deck.

```python
import zoomdeck_nb
```

## One slide

A deck is a list of `::: slide` containers. Every slide has an `#id`; the first heading of a slide is its title in the outline. This deck has one slide and nothing else.

```python
%%deck
::: slide {#start}
# A first deck

Written in Markdown, played in the browser.
:::
```

## Pages, one after another

Two more slides make a flat document: the slides that nothing includes are visited first to last, and the player cuts from one to the next. Press the right arrow on the deck below.

```python
%%deck
::: slide {#start}
# A first deck

Written in Markdown, played in the browser.
:::

::: slide {#second}
## A second page

Plain prose, **bold**, *italic*, `code`, and a formula: $e^{i\pi} + 1 = 0$.
:::

::: slide {#third}
## A third page

- A list
- of points
:::
```

## Steps

`{#id step=N}` on an element reveals it on the N-th click. Elements without `step` are there from the start.

```python
%%deck
::: slide {#steps width=20}
## Steps

[Shown on the first click]{#a step=1}

[And on the second]{#b below:#a left-aligned:#a step=2}

[Then a formula: $\int_0^1 x^2\,dx = \tfrac13$]{#c below:#b left-aligned:#b step=3}
:::
```

## A child slide

An include places another slide inside this one, as a viewport. The camera zooms into the child when its turn comes and back out after the last child. `preview=true` shows the child's content before the visit, so the parent displays it as a tile.

```python
%%deck
::: slide {#start width=20}
# A first deck

[The details are in the child slide, on the right.]{#intro x=-9 y=1}

[[rectangle: #frame x=4.5 y=-1.5 width=9 height=5.06 color=GRAY stroke_width=1]]

[[include: #details x=4.5 y=-1.5 width=9 preview=true]]
:::

::: slide {#details}
## Details

[Shown on the first click]{#a step=1}

[And on the second]{#b step=2}

Elements without a position flow from the top, one under the other. {color=GRAY}
:::
```

## A diagram and an animation

Shapes are directives: an arrow between two elements, a box around one. `[[animate: ...]]` changes an element on a step, here its size and colour; `enter=write` writes text stroke by stroke, here in the bundled handwriting face (`font=Zeyada`).

```python
%%deck
::: slide {#diagram width=20}
## A small diagram

[Input]{#in x=-7 y=1 font_size=28}

[Output]{#out x=4 y=1 font_size=28}

[[arrow: #in -> #out, #flow, label="process", step=1]]

[[box: #in, #b1, buff=0.25, color=TEAL, step=2]]

[[animate: #out, scale=1.3, color=TEAL, step=3]]

[Written by hand]{#hand x=-7 y=-3 font=Zeyada font_size=44 enter=write step=4}
:::
```
